```python
missing_percentage = (df.isnull().sum()/len(df))*100
print("Missing percentage per column")
print(missing_percentage)
```

```python
df = df.drop(columns=df.columns[(df.isnull().mean() * 100) > 40])
df.columns
```

```python
thresold=40
df = df.loc[:,missing_percentage<thresold]
print("Missing Column:")
print(df.columns)
print("new dataset:",df.shape)
```

```python
for col in df.select_dtypes(include='number').columns:
    missing = df[col].isnull().mean() * 100
    if 10 <= missing <= 40:
        print("Imputing:", col)
        df[col] = df[col].fillna(df[col].median())
```

```python
cols_to_impute = [col for col in df.select_dtypes(include='number').columns
                  if 10 <= (df[col].isnull().mean() * 100) <= 40]
```

```python
num_cols = df.select_dtypes(include='number').columns

for col in num_cols:
    missing = df[col].isnull().mean() * 100
    if 10 <= missing <= 40:
        df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum())
```

```python
numeric_cols = df.select_dtypes(include='number').columns
missing_percentage = df[numeric_cols].isnull().mean() * 100
cols_to_impute = missing_percentage[(missing_percentage > 0) & (missing_percentage < 10)].index
print("Columns to impute:")
print(cols_to_impute)

df[cols_to_impute] = df[cols_to_impute].fillna(df[cols_to_impute].mean())
df
```

```python
cols_to_impute = missing_percentage[(missing_percentage >= 10) & (missing_percentage <= 40)].index
print("Columns to impute:")
print(cols_to_impute)

df[cols_to_impute] = df[cols_to_impute].fillna(df[cols_to_impute].median())
df
```

```python
df = df.drop(columns=df.columns[(df.isnull().mean()*100) > 40])
```

```python
num_columns = df.select_dtypes(include=["number"]).columns
missing_percent = (df[num_columns].isna().sum() / len(df)) * 100
mean_cols = []

for col in missing_percent.index:
    if 0 < missing_percent[col] < 10:
        mean_cols.append(col)

print("Columns filled using Mean:")
print(mean_cols)
```

```python
for col in mean_cols:
    df[col] = df[col].fillna(df[col].mean())

df.head()
```

```python
median_cols = []

for col in missing_percent.index:
    if 10 <= missing_percent[col] <= 40:
        median_cols.append(col)

print("Columns filled using Median:")
print(median_cols)
for col in median_cols:
    df[col] = df[col].fillna(df[col].median())

df.head()
```

```python
print("Before Imputation:")
print(df_before.isnull().sum())

print("\n After Imputation:")
print(df.isnull().sum())
```
